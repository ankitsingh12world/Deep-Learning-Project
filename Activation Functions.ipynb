Sigmoid Activation Function

### The sigmoid function is an activation function used in neural networks. It converts any real number into a value between 0 and 1.

```python
def sigmoid(X):
    return 1 / (1 + math.exp(-X))
```

```python
sigmoid(100)
```

```python
sigmoid(1)
```

```python
sigmoid(-56)
```

```python
sigmoid(0.5)
```

Tanh Activation Function

#### The tanh (hyperbolic tangent) is another activation function used in neural networks. It converts input values into a range between −1 and 1.

# tanh

```python
def tanh(x):
    return(math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x))
```

```python
tanh(-56)
```

```python
tanh(1)
```

## ReLU keeps positive values and converts negative values to zero.

```python
def relu(x):
    return max(0,x)
```

```python
relu(-100)
```

------------------------------------------------------------

#### Leaky ReLU keeps positive values same and allows a small negative slope instead of zero.

```python
def leaky_relu(x):
    return max (0.1 *x,x)
```

```python
leaky_relu(-100)
```

```python
leaky_relu(8)
```
