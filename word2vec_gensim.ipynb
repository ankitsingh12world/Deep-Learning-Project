### Gensim is a Python library used for NLP tasks like word embeddings and topic modeling. It helps in converting text into meaningful vectors and finding relationships between words

```python
from gensim.models import Word2Vec

sentences = [["i", "love", "ai"], ["ai", "is", "powerful"]]

model = Word2Vec(sentences, vector_size=100, window=5, min_count=1)

print(model.wv["ai"])  # vector of word "ai"
```

### Simple Preprocessing & Tokenization

 The first thing to do for any data science task is to clean the data. For NLP, we apply various processing like converting all the words to lower case, trimming spaces, removing punctuations. This is something we will do over here too.

Additionally, we can also remove stop words like 'and', 'or', 'is', 'the', 'a', 'an' and convert words to their root forms like 'running' to 'run'.



```python
df.reviewText
```

```python
df.reviewText[0]
```

```python
gensim.utils.simple_preprocess("They look good and stick good! I just don't like the rounded shape because I was always bumping it and Siri kept popping up and it was irritating. I just won't buy a product like this again")
```

```python
review_text = df.reviewText.apply(gensim.utils.simple_preprocess)
```

```python
review_text[0]
```

```python
review_text[1]
```

```python
model = gensim.models.Word2Vec(
    window = 10,
    min_count = 2,
    workers = 4
)
```

build_vocab is used in gensim to create a vocabulary from text data by identifying unique words and their frequency. It prepares the data before training the Word2Vec model.”

#### Build Vocabulary

```python
model.build_vocab(review_text,progress_per=1000)
```

```python
model.epochs
```

#### Epochs is the number of complete passes of the training data through the model.

#### Train the Word2Vec Model

```python
model.train(review_text,total_examples=model.corpus_count,epochs = model.epochs)
```

#### Save the Model

```python
model.save("./word2vec-amazon-cell-accessories-reviews-short.model")
```

#### Finding Similar Words and Similarity between words

```python
model.wv.most_similar("bad")
```

```python
model.wv.similarity(w1="great", w2="good")
```
